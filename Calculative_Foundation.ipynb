```python

# ==========================================================
# PROJECT: CALCULATIVE FOUNDATION
# Topic: Linear Algebra using Students' Performance Scores
# ==========================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ----------------------------------------------------------
# DATASET: STUDENTS' PERFORMANCE SCORES
# ----------------------------------------------------------

data = {
    'Student': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan', 'Riya'],
    'Maths': [85, 72, 95, 60, 88, 76],
    'Science': [78, 90, 85, 70, 92, 82],
    'English': [92, 80, 89, 68, 95, 78],
    'Computer': [88, 75, 93, 72, 90, 85]
}

df = pd.DataFrame(data)

print("STUDENTS PERFORMANCE DATASET")
print(df.to_string(index=False))

# Convert subject scores into a NumPy matrix
students = df['Student'].values
subjects = ['Maths', 'Science', 'English', 'Computer']

A = df[subjects].values

print("\nSTUDENT SCORE MATRIX:")
print(A)


# ==========================================================
# PART A: VECTOR AND MATRIX FUNDAMENTALS
# ==========================================================

print("\n========== PART A: VECTOR OPERATIONS ==========")

# 1. Represent each student's scores as a vector

print("\n1. STUDENT SCORE VECTORS")

for i in range(len(students)):
    print(students[i], "=", A[i])


# 2. Calculate Norm-1 and Norm-2

print("\n2. NORM-1 AND NORM-2")

for i in range(len(students)):
    norm1 = np.linalg.norm(A[i], ord=1)
    norm2 = np.linalg.norm(A[i], ord=2)

    print(students[i])
    print("Norm-1:", norm1)
    print("Norm-2:", round(norm2, 2))


# 3. Dot Product and Angle Between Two Students

print("\n3. DOT PRODUCT AND ANGLE")

v1 = A[0]  # Amit
v2 = A[1]  # Priya

dot_product = np.dot(v1, v2)

cos_theta = dot_product / (
    np.linalg.norm(v1) * np.linalg.norm(v2)
)

cos_theta = np.clip(cos_theta, -1, 1)

angle = np.degrees(np.arccos(cos_theta))

print("Amit's Vector:", v1)
print("Priya's Vector:", v2)
print("Dot Product:", dot_product)
print("Angle Between Vectors:", round(angle, 2), "degrees")


# 4. Cross Product (Using 3 Subjects)

print("\n4. CROSS PRODUCT")

v3 = A[0][:3]  # Amit's first 3 subjects
v4 = A[1][:3]  # Priya's first 3 subjects

cross_product = np.cross(v3, v4)

print("Amit's 3D Vector:", v3)
print("Priya's 3D Vector:", v4)
print("Cross Product:", cross_product)


# 5. Vector Projection

print("\n5. VECTOR PROJECTION")

# Projection of Priya's vector onto Amit's vector

projection = (
    np.dot(v2, v1) / np.dot(v1, v1)
) * v1

print("Original Vector (Priya):", v2)
print("Reference Vector (Amit):", v1)
print("Projection:", np.round(projection, 2))


# ==========================================================
# PART B: MATRIX OPERATIONS
# ==========================================================

print("\n========== PART B: MATRIX OPERATIONS ==========")

# 6. Matrix Addition

print("\n6. MATRIX ADDITION")

matrix_addition = A + A

print(matrix_addition)


# 7. Matrix Multiplication

print("\n7. MATRIX MULTIPLICATION")

# Student-to-student similarity matrix

matrix_multiplication = np.dot(A, A.T)

print(matrix_multiplication)

# Subject-to-subject relationship matrix

subject_multiplication = np.dot(A.T, A)

print("\nSUBJECT RELATIONSHIP MATRIX:")
print(subject_multiplication)


# 8. Matrix Transpose

print("\n8. MATRIX TRANSPOSE")

transpose = A.T

print(transpose)


# 9. Matrix Inverse

print("\n9. MATRIX INVERSE")

# A is rectangular (6 x 4), so its inverse does not exist.
# Use the square subject Gram matrix instead.

square_matrix = A.T @ A

if np.linalg.det(square_matrix) != 0:
    inverse = np.linalg.inv(square_matrix)

    print("Inverse of A.T @ A:")
    print(np.round(inverse, 4))
else:
    print("Inverse does not exist because determinant is zero.")


# 10. Determinant

print("\n10. MATRIX DETERMINANT")

# Select first 4 students and 4 subjects to create a 4 x 4 matrix

square_matrix_2 = A[:4, :4]

determinant = np.linalg.det(square_matrix_2)

print("4 x 4 Matrix:")
print(square_matrix_2)

print("Determinant:", round(determinant, 2))


# ==========================================================
# PART C: LINEAR TRANSFORMATIONS AND GEOMETRY
# ==========================================================

print("\n========== PART C: LINEAR TRANSFORMATIONS ==========")

# 11. Explain Line, Plane and Hyperplane

print("\n11. LINE, PLANE AND HYPERPLANE")

print("""
LINE:
A line is a one-dimensional space.
Example: Student scores represented using one subject.

PLANE:
A plane is a two-dimensional space.
Example: Maths and Science scores.

HYPERPLANE:
A hyperplane is a flat surface in higher dimensions.
For n dimensions, a hyperplane has dimension n-1.

Example:
2D data = 1D line
3D data = 2D plane
4D data = 3D hyperplane
""")

# 12. Show dimensionality from 2D to 3D to higher dimensions

print("\n12. DIMENSIONALITY VISUALIZATION")

# 2D: Maths vs Science

plt.figure(figsize=(7, 5))
plt.scatter(A[:, 0], A[:, 1])

for i, name in enumerate(students):
    plt.annotate(name, (A[i, 0], A[i, 1]))

plt.xlabel("Maths Marks")
plt.ylabel("Science Marks")
plt.title("2D: Student Performance")
plt.grid(True)
plt.show()


# 3D: Maths, Science and English

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(A[:, 0], A[:, 1], A[:, 2])

for i, name in enumerate(students):
    ax.text(A[i, 0], A[i, 1], A[i, 2], name)

ax.set_xlabel("Maths")
ax.set_ylabel("Science")
ax.set_zlabel("English")
ax.set_title("3D: Student Performance")

plt.show()


# Higher Dimensions: Four subjects
# Represented using a 4D data matrix

print("\n4D DATA REPRESENTATION:")
print("Dimensions:", A.shape)
print("Students:", A.shape[0])
print("Subjects:", A.shape[1])
print("Each student is represented by 4 coordinates.")


# ==========================================================
# PART D: EIGENVALUES AND DECOMPOSITION
# ==========================================================

print("\n========== PART D: EIGENVALUES AND DECOMPOSITION ==========")

# 13. Calculate Covariance Matrix

print("\n13. COVARIANCE MATRIX")

covariance_matrix = np.cov(A, rowvar=False)

print(covariance_matrix)


# 14. Calculate Eigenvalues and Eigenvectors

print("\n14. EIGENVALUES AND EIGENVECTORS")

eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)

# Sort eigenvalues in descending order

idx = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

print("\nEIGENVALUES:")
print(np.round(eigenvalues, 2))

print("\nEIGENVECTORS:")
print(np.round(eigenvectors, 4))


# 15. Interpret Eigenvalues

print("\n15. VARIANCE EXPLAINED BY EACH COMPONENT")

variance_percentage = (
    eigenvalues / np.sum(eigenvalues)
) * 100

for i, value in enumerate(variance_percentage):
    print(
        "Principal Component", i + 1,
        ":", round(value, 2), "%"
    )


# 16. Visualize Eigenvalues

plt.figure(figsize=(7, 5))

plt.bar(
    ['PC1', 'PC2', 'PC3', 'PC4'],
    variance_percentage
)

plt.xlabel("Principal Components")
plt.ylabel("Variance Explained (%)")
plt.title("Eigenvalues - Variance Explained")
plt.grid(axis='y')

plt.show()


# ==========================================================
# FINAL PROJECT SUMMARY
# ==========================================================

print("\n========== PROJECT SUMMARY ==========")

print("1. Student scores represented using vectors.")
print("2. Norm-1 and Norm-2 calculated.")
print("3. Dot product and angle calculated.")
print("4. Cross product calculated using 3 subjects.")
print("5. Vector projection calculated.")
print("6. Matrix addition and multiplication performed.")
print("7. Matrix transpose and inverse calculated.")
print("8. Determinant calculated.")
print("9. Line, plane and hyperplane explained.")
print("10. 2D and 3D visualizations created.")
print("11. Covariance matrix calculated.")
print("12. Eigenvalues and eigenvectors calculated.")
print("13. Variance explained visualized.")
```
